## Feature engineering : forme des équipes (rolling stats)

Objectif : pour chaque match, donner au modèle des indicateurs de forme de
chaque équipe **calculés uniquement à partir de ses matchs précédents**
(jamais le match en cours, jamais un match futur — sinon fuite de données :
le modèle "verrait" le résultat qu'il doit prédire).

Étapes :
1. Transformer `df_all` (1 ligne = 1 match) en `df_team_matches`
   (1 ligne = 1 équipe pour 1 match) — chaque match donne donc 2 lignes
   (perspective domicile + perspective extérieur).
2. Trier chronologiquement par équipe, et calculer des moyennes glissantes
   (buts, tirs, corners, points...) sur les 5 et 10 derniers matchs, en
   décalant d'un cran (`shift(1)`) pour exclure le match du jour.
3. Recoller ces stats de forme sur `df_all`, une fois côté équipe à domicile,
   une fois côté équipe à l'extérieur.

```python
import numpy as np

def build_team_perspective(df):
    """Transforme la base 1-ligne-par-match en 1-ligne-par-équipe-par-match."""
    home = pd.DataFrame({
        "date": df["date"],
        "league": df["league"],
        "season": df["season"],
        "team": df["home_team"],
        "opponent": df["away_team"],
        "is_home": True,
        "goals_for": df["full_time_home_goals"],
        "goals_against": df["full_time_away_goals"],
        "shots_for": df["home_shots"],
        "shots_against": df["away_shots"],
        "shots_on_target_for": df["home_shots_on_target"],
        "shots_on_target_against": df["away_shots_on_target"],
        "corners_for": df["home_corners"],
        "corners_against": df["away_corners"],
        "fouls_for": df["home_fouls"],
        "fouls_against": df["away_fouls"],
    })
    away = pd.DataFrame({
        "date": df["date"],
        "league": df["league"],
        "season": df["season"],
        "team": df["away_team"],
        "opponent": df["home_team"],
        "is_home": False,
        "goals_for": df["full_time_away_goals"],
        "goals_against": df["full_time_home_goals"],
        "shots_for": df["away_shots"],
        "shots_against": df["home_shots"],
        "shots_on_target_for": df["away_shots_on_target"],
        "shots_on_target_against": df["home_shots_on_target"],
        "corners_for": df["away_corners"],
        "corners_against": df["home_corners"],
        "fouls_for": df["away_fouls"],
        "fouls_against": df["home_fouls"],
    })

    team_matches = pd.concat([home, away], ignore_index=True)

    # Résultat du point de vue de l'équipe -> points (victoire=3, nul=1, défaite=0)
    conditions = [
        team_matches["goals_for"] > team_matches["goals_against"],
        team_matches["goals_for"] == team_matches["goals_against"],
    ]
    team_matches["points"] = np.select(conditions, [3, 1], default=0)

    team_matches = team_matches.sort_values(["team", "date"]).reset_index(drop=True)
    return team_matches

df_team_matches = build_team_perspective(df_all)
print(f"{df_team_matches.shape[0]} lignes (2 x {df_all.shape[0]} matchs)")
df_team_matches.head()
```

```python
def add_rolling_features(df, group_col="team", windows=(5, 10)):
    """
    Calcule des moyennes glissantes par équipe, en n'utilisant que les matchs
    STRICTEMENT antérieurs (shift(1)) pour éviter toute fuite de données.
    """
    df = df.sort_values([group_col, "date"]).copy()
    grouped = df.groupby(group_col, group_keys=False)

    stat_cols = [
        "goals_for", "goals_against", "shots_for", "shots_against",
        "shots_on_target_for", "shots_on_target_against",
        "corners_for", "corners_against", "fouls_for", "fouls_against", "points",
    ]

    for window in windows:
        for col in stat_cols:
            new_col = f"{col}_avg_last{window}"
            df[new_col] = grouped[col].transform(
                lambda s: s.shift(1).rolling(window, min_periods=1).mean()
            )

    # Nombre de matchs déjà joués avant celui-ci (utile pour filtrer les débuts
    # de saison / débuts d'historique, où la forme calculée est peu fiable)
    df["matches_played_before"] = grouped.cumcount()

    return df

df_team_matches = add_rolling_features(df_team_matches)
df_team_matches[
    ["team", "date", "goals_for", "goals_for_avg_last5", "points_avg_last5", "matches_played_before"]
].head(15)
```

### Forme spécifique domicile / extérieur

Les rolling stats ci-dessus mélangent, pour chaque équipe, ses matchs à
domicile et à l'extérieur dans une seule moyenne — on perd l'avantage du
terrain (une équipe peut être moyenne en général mais très forte chez elle).

On calcule donc les mêmes moyennes glissantes, mais séparément pour les
matchs à domicile et les matchs à l'extérieur de chaque équipe
(`groupby(["team", "is_home"])` au lieu de `groupby("team")`). Ainsi, côté
domicile du match, on utilisera la forme de l'équipe **spécifiquement à
domicile** ; côté extérieur, sa forme **spécifiquement à l'extérieur**.

```python
def add_rolling_features_by_venue(df, windows=(5, 10)):
    """
    Comme add_rolling_features, mais la moyenne glissante est calculée
    séparément selon que l'équipe jouait à domicile ou à l'extérieur.
    Colonnes produites : "<stat>_avg_last<window>_venue".
    """
    df = df.sort_values(["team", "is_home", "date"]).copy()
    grouped = df.groupby(["team", "is_home"], group_keys=False)

    stat_cols = [
        "goals_for", "goals_against", "shots_for", "shots_against",
        "shots_on_target_for", "shots_on_target_against",
        "corners_for", "corners_against", "fouls_for", "fouls_against", "points",
    ]

    for window in windows:
        for col in stat_cols:
            new_col = f"{col}_avg_last{window}_venue"
            df[new_col] = grouped[col].transform(
                lambda s: s.shift(1).rolling(window, min_periods=1).mean()
            )

    return df

df_team_matches = add_rolling_features_by_venue(df_team_matches)
df_team_matches[
    ["team", "is_home", "date", "goals_for", "goals_for_avg_last5", "goals_for_avg_last5_venue"]
].head(15)
```

### Repos entre matchs

Nombre de jours écoulés depuis le match précédent de l'équipe (toutes
compétitions confondues — championnat + Champions League). Capture la
fatigue / les enchaînements rapprochés (ex: match de Ligue des Champions en
semaine suivi d'un match de championnat le week-end).

```python
def add_rest_days(df):
    """Jours depuis le match précédent de l'équipe, toutes compétitions confondues."""
    df = df.sort_values(["team", "date"]).copy()
    df["days_since_last_match"] = df.groupby("team")["date"].diff().dt.days
    return df

df_team_matches = add_rest_days(df_team_matches)
df_team_matches[["team", "date", "days_since_last_match"]].head(10)
```

### Historique des confrontations directes (head-to-head)

Comment une équipe s'en sort habituellement **contre cet adversaire précis**
(certains matchs ont des dynamiques récurrentes indépendantes de la forme du
moment — rivalités, styles de jeu qui s'affrontent mal, etc.).

Les confrontations directes étant rares (2 fois par saison en championnat,
parfois aucune en Champions League d'une année sur l'autre), une fenêtre
glissante de 5 matchs n'aurait presque jamais assez de données. On utilise
donc une **moyenne expanding** (sur tout l'historique disponible, aussi
petit soit-il) plutôt qu'une moyenne glissante à fenêtre fixe.

```python
def add_head_to_head_features(df):
    """
    Historique des confrontations directes entre une équipe et un adversaire
    précis : moyenne expanding (tout l'historique dispo) des points obtenus
    et de la différence de buts, toujours décalée d'un cran (shift(1)) pour
    exclure le match du jour.
    """
    df = df.sort_values(["team", "opponent", "date"]).copy()
    grouped = df.groupby(["team", "opponent"], group_keys=False)

    df["goal_diff"] = df["goals_for"] - df["goals_against"]

    df["h2h_points_avg"] = grouped["points"].transform(lambda s: s.shift(1).expanding().mean())
    df["h2h_goal_diff_avg"] = grouped["goal_diff"].transform(lambda s: s.shift(1).expanding().mean())
    df["h2h_matches_played"] = grouped.cumcount()

    return df

df_team_matches = add_head_to_head_features(df_team_matches)
df_team_matches[
    ["team", "opponent", "date", "points", "h2h_points_avg", "h2h_matches_played"]
].head(10)
```

### Rating Elo par équipe

Un rating unique par équipe qui résume sa force globale, mis à jour après
**chaque** match (gain de rating contre une équipe plus forte que prévu,
perte contre une équipe plus faible que prévu). Contrairement aux moyennes
glissantes, il n'y a jamais de `NaN` : toute équipe démarre avec un rating
par défaut (1500) et il évolue ensuite match après match.

Le calcul est intrinsèquement **séquentiel** (le rating après le match N
dépend du rating après le match N-1) — on ne peut pas le vectoriser avec
`groupby`/`rolling` comme les autres features, on fait donc une boucle
`for` classique sur les matchs triés par date. On enregistre le rating
**avant** le match comme feature (donc sans fuite de données), puis on met
à jour les deux ratings avec le résultat réel.

```python
def compute_elo_ratings(df, k=20, initial_rating=1500, home_advantage=100):
    """
    Calcule un rating Elo par équipe, mis à jour après chaque match.
    k : vitesse d'ajustement du rating (plus haut = plus réactif, plus bruité).
    home_advantage : bonus (en points Elo) accordé à l'équipe à domicile
    avant de calculer sa probabilité de victoire attendue.
    """
    df = df.sort_values("date").reset_index(drop=True).copy()
    ratings = {}

    home_elo_before = []
    away_elo_before = []

    for _, row in df.iterrows():
        home, away = row["home_team"], row["away_team"]
        home_rating = ratings.get(home, initial_rating)
        away_rating = ratings.get(away, initial_rating)

        home_elo_before.append(home_rating)
        away_elo_before.append(away_rating)

        # Probabilité de victoire attendue (formule Elo standard)
        expected_home = 1 / (1 + 10 ** ((away_rating - (home_rating + home_advantage)) / 400))

        if row["full_time_home_goals"] > row["full_time_away_goals"]:
            actual_home = 1.0
        elif row["full_time_home_goals"] == row["full_time_away_goals"]:
            actual_home = 0.5
        else:
            actual_home = 0.0

        ratings[home] = home_rating + k * (actual_home - expected_home)
        ratings[away] = away_rating + k * ((1 - actual_home) - (1 - expected_home))

    df["home_elo_before"] = home_elo_before
    df["away_elo_before"] = away_elo_before

    return df

df_all = compute_elo_ratings(df_all)
df_all[["date", "home_team", "away_team", "home_elo_before", "away_elo_before"]].head(10)
```

### Recollage des features sur la base match par match

```python
# On récupère uniquement les colonnes de features glissantes (pas les stats brutes du match)
feature_cols = [c for c in df_team_matches.columns if "_avg_last" in c] + [
    "matches_played_before", "days_since_last_match",
    "h2h_points_avg", "h2h_goal_diff_avg", "h2h_matches_played",
]

home_features = df_team_matches.loc[df_team_matches["is_home"], ["date", "team"] + feature_cols].copy()
home_features = home_features.rename(columns={"team": "home_team", **{c: f"home_{c}" for c in feature_cols}})

away_features = df_team_matches.loc[~df_team_matches["is_home"], ["date", "team"] + feature_cols].copy()
away_features = away_features.rename(columns={"team": "away_team", **{c: f"away_{c}" for c in feature_cols}})

df_features = df_all.merge(home_features, on=["date", "home_team"], how="left")
df_features = df_features.merge(away_features, on=["date", "away_team"], how="left")

print(f"{df_features.shape[0]} matchs, {df_features.shape[1]} colonnes")
df_features.filter(regex="date|home_team|away_team|goals_for_avg|points_avg|matches_played|days_since|h2h").head()
```
